```python
students=['Ilya','Galina', 'Timur','Vasya','Katya']
```

```python
def get_students():
    """ Этот код описывает поведение функции get_students()
мы ищем студентов с длиной имени больше 5 букв
"""
    for i in students:
        if len(i)>5:
            print(i)
    return 'Успешно'
```

```python
get_students()
```
