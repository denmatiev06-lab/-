```python
students=["Alisa","Denis","Anton","evgenyi","Sergey","Timyr","Katya","Vasilyi","Linda","Matvei"]
def replace_vowel_names(students_list):
    vowels="AEIOUY"
    return ["invalid" if name [0] in vowels else name for name in students_list]
students=replace_vowel_names(students)
print(students)

def count_invalid(lst):
    return lst.count("invalid")
invalid_count=count_invalid(students)
print(f"{invalid_count}")
```
