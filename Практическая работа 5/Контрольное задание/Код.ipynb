```python
import os
for root, dirs, files in os.walk(os.path.abspath(os.sep)):
    if 'titanik.csv' in files:
        file_path = os.path.join(root, 'titanik.csv')
        break
else:
    print("Файл titanik.csv не найден.")
    file_path = None
if file_path:
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()
        for line in lines[-3:]:
            print(line, end='')
```
