```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
file_path='titanic.csv'
df=pd.read_csv(file_path)
df.columns=[c.lower().replace(' ','_') for c in df.columns]
df_age=df[(df['age']>=20)&(df['age']<=30)]
fare_stats={
    'average':df_age['fare'].mean(),
    'median': df_age['fare'].median(),
    'min':df_age['fare'].min(),
    'max':df_age['fare'].max()
}
print('Статистичекие данные:')
for k,v in fare_stats.items():
    print(f'{k}:[v:.2f]')
# График
plt.figure(figsize=(8,6))
sns.violinplot(x='sex',y='fare',data=df_age)
plt.title('График')
plt.xlabel('Пол')
plt.ylabel('Цена билета')
plt.tight_layout()
plt.show()
```
